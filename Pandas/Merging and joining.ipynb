```python
import pandas as pd
employees = pd.DataFrame({
    "EmpID": [1, 2, 3],
    "Name": ["Alice", "Bob", "Charlie"],
    "DeptID": [10, 20, 30]
})

departments = pd.DataFrame({
    "DeptID": [10, 20, 40],
    "DeptName": ["HR", "Engineering", "Marketing"]
})
```

```python
employees
```

```python
departments
```

```python
pd.merge(employees, departments, on="DeptID")#inner join common thhings in both table
```

```python
pd.merge(employees, departments, on="DeptID",how="left")#ileft join lefty table completely incliuded
```

```python
pd.merge(employees, departments, on="DeptID",how="right")#right join right table completely incliuded
```

```python
pd.merge(employees, departments, on="DeptID",how="outer")#outer join both table completely incliuded
```
